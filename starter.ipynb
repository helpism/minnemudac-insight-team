# Load all files

```python
import os
import pandas as pd

pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)

organization_path = "data/Organization.xlsx"
marketbucks_path = "data/marketbucksUsage.xlsx"
call_path = "data/Incoming Call.xlsx"
food_path = "data/org_FSStats.xlsx"

organization_records = pd.read_excel(organization_path)
marketbucks_participation = pd.read_excel(marketbucks_path)
call_center_logs = pd.read_excel(call_path)
food_shelf_activity = pd.read_excel(food_path)
```

## Row and Column Count

```python
# How many rows and columns in all files
print("Number of rows in Organization:", len(organization_records))
print("Number of columns in Organization:", len(organization_records.columns))
print()
print("Number of rows in marketbucksUsage:", len(marketbucks_participation))
print("Number of columns in marketbucksUsage:", len(marketbucks_participation.columns))
print()
print("Number of rows in Incoming Call:", len(call_center_logs))
print("Number of columns in Incoming Call:", len(call_center_logs.columns))
print()
print("Number of rows in org_FSStats:", len(food_shelf_activity))
print("Number of columns in org_FSStats:", len(food_shelf_activity.columns))
```
